# 6.3 中心化协同：一个指挥官控制所有无人机

> **AirSim 配置**：使用 `1-6-settings.json`（2架无人机）。如已在运行笔记本2，无需重启。

## 核心思想

中心化协同是最直观的多无人机协作模式：

```
人类操作员  →  [指挥官LLM]  →  分解任务  →  逐一指挥每架无人机执行
```

一个 LLM 充当"指挥官"，接收高层任务目标，将其分解为具体的飞行指令，然后依次控制每架无人机完成任务。

**整个实现不到50行代码。**

## Step 1：导入工具

````python
import json
from airsim_tools import connect_airsim, takeoff, fly_to, get_state, call_llm
````

## Step 2：连接 AirSim

````python
client = connect_airsim()
client.reset()
````

## Step 3：指挥官 LLM 分解任务

给 LLM 一个高层任务目标，让它自动分解为每架无人机的具体航点任务。

LLM 返回 JSON 格式，我们直接解析执行。

````python
# ==================== 中心化协同核心代码 ====================

# 1. 定义任务目标
mission = "派Drone1去(-10, 10, -10)侦察，派Drone2去(10, 10, -10)侦察"

# 2. 指挥官 LLM 分解任务
plan_prompt = f"""你是无人机集群指挥官。请将以下任务分解为具体飞行指令。

任务：{mission}

要求：
- 输出纯JSON列表，不要其他文字
- 每个任务格式：{{"drone_id": "Drone1", "target": [x, y, z]}}
- z坐标为负表示高度（如-10表示10米高）
"""

plan_text = call_llm(plan_prompt)
print("指挥官规划结果：")
print(plan_text)

# 解析 JSON
tasks = json.loads(plan_text)
print(f"\n共 {len(tasks)} 个任务")
````

## Step 4：顺序执行所有任务

指挥官分配好了任务，现在逐一执行：起飞 → 飞往目标 → 记录结果。

````python
# 3. 逐一执行
results = []

for i, task in enumerate(tasks):
    drone_id = task["drone_id"]
    target = task["target"]  # [x, y, z]
    
    print(f"\n--- 执行任务 {i+1}: {drone_id} → {target} ---")
    
    # 起飞
    takeoff(client, drone_id)
    
    # 飞往目标
    fly_to(client, drone_id, target[0], target[1], target[2])
    
    # 记录位置
    pos = get_state(client, drone_id)
    results.append({"drone_id": drone_id, "target": target, "actual_pos": pos})
    print(f"[{drone_id}] 实际位置: {pos}")

print("\n所有任务执行完毕！")
````

## Step 5：指挥官 LLM 生成总结报告

所有任务完成后，让 LLM 生成一份简洁的执行报告。

````python
# 4. LLM 总结
summary_prompt = f"""以下是无人机任务执行结果，请生成简洁的中文报告：

原始任务：{mission}
执行结果：{json.dumps(results, ensure_ascii=False)}

请分析每架无人机是否到达目标位置，总结任务完成情况。
"""

report = call_llm(summary_prompt)
print("===== 任务报告 =====")
print(report)
````

## 流程总结

中心化协同的完整流程只有4步：

```
1. 人类输入任务目标
      ↓
2. LLM指挥官 → 分解为JSON任务列表
      ↓
3. for循环 → 逐一执行（takeoff + fly_to）
      ↓
4. LLM指挥官 → 生成总结报告
```

**优点**：
- 代码极简，一目了然
- 指挥官有全局信息，可以做出全局最优分配
- 适合任务明确、通信可靠的场景

**局限**：
- 指挥官是单点故障
- 无人机之间没有直接通信
- 所有决策集中在一个LLM上，无法应对动态变化
