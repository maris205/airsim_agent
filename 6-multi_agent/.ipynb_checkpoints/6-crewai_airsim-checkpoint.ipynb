# 6.6 CrewAI + AirSim：多无人机协同任务

> **AirSim 配置**：使用 `1-6-settings.json`（2架无人机）。复制到 `~/Documents/AirSim/settings.json` 后启动 AirSim。
>
> **Python 环境**：需要 Python 3.10+，使用 `conda activate drone` 环境。

本节将 CrewAI 框架与 AirSim 模拟器结合，用 Agent 框架编排多无人机协同搜索任务。

核心思路：
- 用 CrewAI 的 `@tool` 将 AirSim 飞行控制封装为 Agent 可调用的工具
- 定义两个 Agent：**任务规划师**（分解任务）和**侦察员**（执行飞行）
- 组建 Crew，一键执行整个协同流程

## Step 1：安装与导入

````python
#!pip install crewai openai airsim
````

````python
import sys
sys.path.append('../external-libraries')

import json
import airsim
from crewai import Agent, Task, Crew, LLM
from crewai.tools import tool
````

## Step 2：配置 LLM 和 AirSim 连接

````python
# 配置豆包 LLM
llm = LLM(
    model="doubao-seed-2-0-pro-260215",
    base_url="https://ark.cn-beijing.volces.com/api/v3",
    api_key="",
    temperature=0.1
)

# 连接 AirSim
airsim_client = airsim.MultirotorClient()
airsim_client.confirmConnection()
airsim_client.reset()
print(f"AirSim 连接成功，无人机: {airsim_client.listVehicles()}")
````

## Step 3：将 AirSim API 封装为 CrewAI 工具

用 `@tool` 装饰器将无人机控制函数包装成 Agent 可调用的工具。Agent 会根据任务需要，**自动决定何时调用哪个工具**。

````python
@tool("drone_takeoff")
def drone_takeoff(drone_id: str) -> str:
    """命令指定的无人机起飞。参数 drone_id 为无人机名称，如 'Drone1'。"""
    airsim_client.enableApiControl(True, vehicle_name=drone_id)
    airsim_client.armDisarm(True, vehicle_name=drone_id)
    airsim_client.takeoffAsync(vehicle_name=drone_id).join()
    return f"{drone_id} 已成功起飞"


@tool("drone_fly_to")
def drone_fly_to(drone_id: str, x: float, y: float, z: float) -> str:
    """命令无人机飞往指定NED坐标。z为负表示高度，如z=-10表示10米高。
    参数：drone_id(无人机名称), x(北方向), y(东方向), z(高度,负数)。"""
    airsim_client.moveToPositionAsync(x, y, z, 5, vehicle_name=drone_id).join()
    return f"{drone_id} 已到达坐标 ({x}, {y}, {z})"


@tool("drone_get_position")
def drone_get_position(drone_id: str) -> str:
    """获取指定无人机的当前位置坐标。参数 drone_id 为无人机名称。"""
    state = airsim_client.getMultirotorState(vehicle_name=drone_id)
    pos = state.kinematics_estimated.position
    return f"{drone_id} 当前位置: x={pos.x_val:.1f}, y={pos.y_val:.1f}, z={pos.z_val:.1f}"


print("AirSim 工具定义完成：drone_takeoff, drone_fly_to, drone_get_position")
````

## Step 4：定义 Agent

两个角色：
- **任务规划师（MissionPlanner）**：负责理解任务目标，分解为具体的飞行指令
- **侦察员（Scout）**：负责执行飞行任务，使用 AirSim 工具控制无人机

````python
# 任务规划师（不需要工具，只负责规划）
planner = Agent(
    role="无人机任务规划师",
    goal="接收高层任务目标，分解为每架无人机的具体飞行坐标",
    backstory=(
        "你是一位经验丰富的无人机任务指挥官。"
        "你擅长将搜索任务分解为多架无人机的具体航点。"
        "你熟悉NED坐标系：z为负表示高度（如z=-10表示10米高）。"
    ),
    llm=llm,
    verbose=True
)

# 侦察员（拥有AirSim飞行工具）
scout = Agent(
    role="无人机侦察员",
    goal="按照规划师的指令，控制无人机执行飞行任务",
    backstory=(
        "你是一名无人机飞行操作员。"
        "你需要按照任务指令，依次让每架无人机起飞并飞往目标坐标。"
        "执行顺序：先起飞（drone_takeoff），再飞往目标（drone_fly_to），最后确认位置（drone_get_position）。"
    ),
    tools=[drone_takeoff, drone_fly_to, drone_get_position],
    llm=llm,
    verbose=True
)

print("Agent 定义完成：planner, scout")
````

## Step 5：定义任务 & 组建 Crew

````python
# 任务目标
mission = "派Drone1飞到(-10, 10, -10)侦察北区，派Drone2飞到(10, 10, -10)侦察南区"

# Task 1：规划任务
plan_task = Task(
    description=f"""将以下任务分解为具体的飞行指令：
    
任务：{mission}

请输出每架无人机的任务列表，包含 drone_id 和目标坐标。""",
    expected_output="每架无人机的具体飞行指令，包含drone_id和target坐标。",
    agent=planner
)

# Task 2：执行飞行
exec_task = Task(
    description="""根据规划师的指令，依次执行每架无人机的飞行任务：

对每架无人机执行以下步骤：
1. 调用 drone_takeoff 让无人机起飞
2. 调用 drone_fly_to 飞往目标坐标
3. 调用 drone_get_position 确认到达位置

请依次完成所有无人机的任务。""",
    expected_output="所有无人机的执行结果报告，包含起飞状态、目标坐标和实际到达位置。",
    agent=scout
)

# 组建 Crew（顺序执行：先规划，再执行）
crew = Crew(
    agents=[planner, scout],
    tasks=[plan_task, exec_task],
    verbose=True
)

print("Crew 组建完成，准备执行")
````

## Step 6：执行！

一行代码启动整个多Agent协同流程。请同时观察 AirSim 模拟器窗口，可以看到无人机起飞和飞行。

````python
# 启动 Crew
result = crew.kickoff()

print("\n" + "="*50)
print("===== 任务执行报告 =====")
print("="*50)
print(result)
````

## Step 7：验证无人机位置

````python
# 手动验证每架无人机的最终位置
for drone_id in ["Drone1", "Drone2"]:
    state = airsim_client.getMultirotorState(vehicle_name=drone_id)
    pos = state.kinematics_estimated.position
    print(f"{drone_id}: x={pos.x_val:.1f}, y={pos.y_val:.1f}, z={pos.z_val:.1f}")
````

## 流程回顾

整个流程由 CrewAI 框架自动编排：

```
crew.kickoff()
  │
  ├─ Task1: 规划师（planner）
  │   └─ LLM思考 → 输出飞行指令
  │
  └─ Task2: 侦察员（scout）
      └─ LLM思考 → 调用 drone_takeoff → 调用 drone_fly_to → 调用 drone_get_position
                    (Agent 自动决定何时调用哪个工具)
```

### 与纯Python实现的对比

| 对比 | 纯Python（笔记本3-4） | CrewAI（本节） |
|------|----------------------|----------------|
| 任务分解 | `call_llm()` + 手动解析JSON | Agent自动推理 |
| 工具调用 | 手动 `takeoff()` + `fly_to()` | Agent根据任务自动选择工具 |
| 流程编排 | for循环/消息板 | Crew自动管理Task顺序 |
| 错误处理 | 需要自己写 | 框架内置重试机制 |
| 代码可读性 | 直接但代码多 | 声明式、更清晰 |
