# 8.0 AirSim 场景数据采集

本 Notebook 在 AirSim 仿真器中采集 VLN（视觉语言导航）训练数据，生成与 UAV-Flow 数据集格式完全一致的数据，可直接用于 OpenVLA 微调。

**采集流程：**
1. 定义导航任务（语言指令 + 航点路径）
2. 无人机按航点飞行，逐帧采集 FPV 图像 + 6-DoF 位姿
3. 计算帧间相对动作（与 UAV-Flow 的 preprocessed_logs 一致）
4. 保存为 UAV-Flow 格式：每条轨迹一个文件夹，含图像序列 + log.json

**生成数据格式：**
```
airsim_vln_data/
├── traj_001/
│   ├── 000000.jpg
│   ├── 000001.jpg
│   ├── ...
│   └── log.json
├── traj_002/
│   └── ...
```

## 0.1 环境准备

````python
import sys
sys.path.append('../external-libraries')
sys.path.append('../airsim_v2')

import airsim
import numpy as np
import cv2
import json
import math
import time
import os
from pathlib import Path
from PIL import Image
from datetime import datetime
import matplotlib.pyplot as plt

%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 6)

# 数据保存路径
OUTPUT_DIR = Path("airsim_vln_data")
OUTPUT_DIR.mkdir(exist_ok=True)

# 图像采集分辨率（与 UAV-Flow 一致）
IMG_SIZE = 224

print(f"数据保存目录: {OUTPUT_DIR.absolute()}")
````

## 0.2 AirSim 连接与工具函数

````python
# 连接 AirSim
client = airsim.MultirotorClient()
client.confirmConnection()
client.enableApiControl(True)
client.armDisarm(True)

def get_pose():
    """获取无人机 6-DoF 位姿 [x, y, z, roll, pitch, yaw]"""
    pose = client.simGetVehiclePose()
    pos = pose.position
    orient = pose.orientation
    pitch, roll, yaw = airsim.to_eularian_angles(orient)
    return [
        pos.x_val, pos.y_val, pos.z_val,
        math.degrees(roll), math.degrees(pitch), math.degrees(yaw)
    ]

def capture_fpv(save_path, img_size=IMG_SIZE):
    """采集前置摄像头 FPV 图像，缩放到指定大小并保存"""
    response = client.simGetImage('0', airsim.ImageType.Scene, vehicle_name='')
    if not response:
        return False
    img = cv2.imdecode(np.array(bytearray(response), dtype='uint8'), cv2.IMREAD_UNCHANGED)
    if img is None:
        return False
    # RGBA -> RGB
    if img.shape[2] == 4:
        img = cv2.cvtColor(img, cv2.COLOR_RGBA2RGB)
    # 缩放到 224x224
    img = cv2.resize(img, (img_size, img_size))
    cv2.imwrite(str(save_path), img)
    return True

print(f"当前无人机位姿: {get_pose()}")
````

## 0.3 定义导航任务

每个任务包含：
- `instruction`: 英文自然语言导航指令（与 UAV-Flow 一致）
- `waypoints`: 无人机要飞过的航点坐标列表 `[x, y, z]`（NED 坐标系）
- `speed`: 飞行速度（m/s）

你可以根据自己的 AirSim 场景修改航点和指令。

````python
# 定义多条导航轨迹任务
# 根据你的场景修改航点坐标和指令
TASKS = [
    {
        "instruction": "Fly forward along the road",
        "waypoints": [[5, 0, -2], [10, 0, -2], [15, 0, -2], [20, 0, -2]],
        "speed": 2.0,
    },
    {
        "instruction": "Turn left and fly toward the building",
        "waypoints": [[5, 0, -2], [8, -3, -2], [10, -8, -2], [10, -12, -2]],
        "speed": 2.0,
    },
    {
        "instruction": "Turn right and go around the house",
        "waypoints": [[5, 0, -2], [8, 3, -2], [10, 8, -3], [8, 12, -3], [5, 12, -2]],
        "speed": 1.5,
    },
    {
        "instruction": "Fly up and then move forward",
        "waypoints": [[0, 0, -5], [0, 0, -8], [5, 0, -8], [10, 0, -8]],
        "speed": 2.0,
    },
    {
        "instruction": "Circle around the tree on the right",
        "waypoints": [[5, 3, -2], [8, 5, -2], [8, 8, -2], [5, 8, -2], [3, 5, -2]],
        "speed": 1.5,
    },
]

print(f"共定义 {len(TASKS)} 条导航任务:")
for i, task in enumerate(TASKS):
    print(f"  [{i}] {task['instruction']}  ({len(task['waypoints'])} 航点, {task['speed']} m/s)")
````

## 0.4 数据采集核心函数

采集过程：
1. 起飞到初始高度
2. 开始逐航点飞行，**飞行过程中**按固定间隔采集 FPV 图像 + 位姿
3. 飞行完成后，计算帧间相对动作（preprocessed_logs）
4. 保存为 UAV-Flow 格式

````python
def compute_relative_actions(raw_logs):
    """
    计算帧间相对动作，与 UAV-Flow 的 preprocessed_logs 格式一致。
    
    输入: raw_logs = [[x, y, z, roll, pitch, yaw, timestamp], ...]
    输出: preprocessed_logs = [[dx, dy, dz, d_roll, d_pitch, d_yaw], ...]
    """
    preprocessed = []
    for i in range(len(raw_logs) - 1):
        curr = raw_logs[i]
        next_ = raw_logs[i + 1]
        
        # 位置差（相对于当前朝向的局部坐标系）
        dx_global = next_[0] - curr[0]
        dy_global = next_[1] - curr[1]
        dz = next_[2] - curr[2]
        
        # 转换到机体坐标系
        yaw_rad = math.radians(curr[5])  # 当前 yaw
        dx_local = dx_global * math.cos(yaw_rad) + dy_global * math.sin(yaw_rad)
        dy_local = -dx_global * math.sin(yaw_rad) + dy_global * math.cos(yaw_rad)
        
        # 姿态差
        d_roll = next_[3] - curr[3]
        d_pitch = next_[4] - curr[4]
        d_yaw = next_[5] - curr[5]
        
        # 处理 yaw 跨越 ±180 度
        if d_yaw > 180:
            d_yaw -= 360
        elif d_yaw < -180:
            d_yaw += 360
        
        preprocessed.append([dx_local, dy_local, dz, d_roll, d_pitch, d_yaw])
    
    # 最后一帧复制倒数第二帧的动作
    if preprocessed:
        preprocessed.append(preprocessed[-1])
    
    return preprocessed


def collect_trajectory(task, traj_id, capture_interval=0.2):
    """
    执行一条导航任务并采集数据。
    
    Args:
        task: 任务字典 {instruction, waypoints, speed}
        traj_id: 轨迹编号
        capture_interval: 采集间隔（秒），默认 0.2s = 5Hz
    """
    traj_name = f"traj_{traj_id:03d}"
    traj_dir = OUTPUT_DIR / traj_name
    traj_dir.mkdir(exist_ok=True)
    
    instruction = task["instruction"]
    waypoints = task["waypoints"]
    speed = task["speed"]
    
    print(f"\n{'='*60}")
    print(f"采集轨迹 [{traj_name}]: {instruction}")
    print(f"航点数: {len(waypoints)}, 速度: {speed} m/s")
    
    # 重置并起飞
    client.reset()
    time.sleep(1)
    client.enableApiControl(True)
    client.armDisarm(True)
    client.takeoffAsync().join()
    time.sleep(0.5)
    
    raw_logs = []
    frame_idx = 0
    
    # 逐航点飞行
    for wp_idx, wp in enumerate(waypoints):
        print(f"  飞往航点 {wp_idx}: {wp}")
        
        # 异步飞行（不 join，这样可以边飞边采集）
        client.moveToPositionAsync(wp[0], wp[1], wp[2], speed)
        
        # 等待到达航点，同时持续采集
        while True:
            # 采集图像
            img_path = traj_dir / f"{frame_idx:06d}.jpg"
            if capture_fpv(img_path):
                # 记录位姿
                pose = get_pose()
                pose.append(time.time())  # 添加时间戳
                raw_logs.append(pose)
                frame_idx += 1
            
            # 检查是否到达航点
            curr_pos = get_pose()[:3]
            dist = math.sqrt(sum((a - b) ** 2 for a, b in zip(curr_pos, wp)))
            if dist < 0.5:  # 到达阈值 0.5 米
                break
            
            time.sleep(capture_interval)
    
    # 计算相对动作
    preprocessed = compute_relative_actions(raw_logs)
    
    # 保存 log.json（与 UAV-Flow 格式一致）
    log_data = {
        "id": traj_name,
        "instruction": instruction,
        "length": frame_idx,
        "raw_logs": [[float(v) for v in row] for row in raw_logs],
        "preprocessed_logs": [[float(v) for v in row] for row in preprocessed],
        "source": "airsim",
        "timestamp": datetime.now().isoformat(),
    }
    
    with open(traj_dir / "log.json", "w") as f:
        json.dump(log_data, f, indent=2)
    
    # 降落
    client.landAsync().join()
    
    print(f"  采集完成! 帧数: {frame_idx}, 保存至: {traj_dir}")
    return log_data

print("采集函数已定义")
````

## 0.5 开始采集

对每个任务执行飞行并采集数据。确保 AirSim 仿真器已启动。

````python
# 执行所有采集任务
all_logs = []

for i, task in enumerate(TASKS):
    log = collect_trajectory(task, traj_id=i+1)
    all_logs.append(log)
    time.sleep(2)  # 任务间隔

print(f"\n{'='*60}")
print(f"全部采集完成! 共 {len(all_logs)} 条轨迹")
print(f"数据目录: {OUTPUT_DIR.absolute()}")
````

## 0.6 数据验证与可视化

验证采集的数据格式是否与 UAV-Flow 一致。

````python
# 验证数据格式
print("数据格式验证:")
print("=" * 50)

for traj_dir in sorted(OUTPUT_DIR.iterdir()):
    if not traj_dir.is_dir():
        continue
    log_path = traj_dir / "log.json"
    if not log_path.exists():
        continue
    
    with open(log_path) as f:
        log = json.load(f)
    
    n_images = len(list(traj_dir.glob("*.jpg")))
    n_raw = len(log["raw_logs"])
    n_pre = len(log["preprocessed_logs"])
    raw_dims = len(log["raw_logs"][0]) if log["raw_logs"] else 0
    pre_dims = len(log["preprocessed_logs"][0]) if log["preprocessed_logs"] else 0
    
    status = "OK" if (n_images == n_raw == n_pre and raw_dims == 7 and pre_dims == 6) else "WARN"
    print(f"  [{status}] {log['id']}: {log['instruction'][:40]}...")
    print(f"       图像={n_images}, raw_logs={n_raw}x{raw_dims}, preprocessed={n_pre}x{pre_dims}")
````

````python
# 可视化一条采集的轨迹
sample_dir = sorted([d for d in OUTPUT_DIR.iterdir() if d.is_dir()])[0]
with open(sample_dir / "log.json") as f:
    log = json.load(f)

# 上排：关键帧图像
images = sorted(list(sample_dir.glob("*.jpg")))
indices = np.linspace(0, len(images)-1, min(6, len(images)), dtype=int)

fig, axes = plt.subplots(2, len(indices), figsize=(3*len(indices), 7))
if len(indices) == 1:
    axes = axes.reshape(2, 1)

fig.suptitle(f"轨迹: {log['id']}\n指令: {log['instruction']}", fontsize=13)

for col, idx in enumerate(indices):
    img = Image.open(images[idx])
    axes[0, col].imshow(img)
    axes[0, col].set_title(f"Frame {idx}", fontsize=9)
    axes[0, col].axis("off")

# 下排：3D 轨迹
raw = np.array(log["raw_logs"])
ax3d = fig.add_subplot(2, 1, 2, projection='3d')
# 隐藏下排的 2D axes
for col in range(len(indices)):
    axes[1, col].set_visible(False)

x, y, z = raw[:, 0] - raw[0, 0], raw[:, 1] - raw[0, 1], raw[:, 2] - raw[0, 2]
ax3d.plot(x, y, z, 'b-o', markersize=2, linewidth=1.5)
ax3d.scatter(x[0], y[0], z[0], color='green', s=80, marker='^', label='Start')
ax3d.scatter(x[-1], y[-1], z[-1], color='red', s=80, marker='*', label='End')
ax3d.set_xlabel('X'); ax3d.set_ylabel('Y'); ax3d.set_zlabel('Z')
ax3d.set_title('3D Flight Path')
ax3d.legend()

plt.tight_layout()
plt.show()
````

````python
# 动作分布（与 01_data_exploration 对比）
all_actions = []
for traj_dir in sorted(OUTPUT_DIR.iterdir()):
    log_path = traj_dir / "log.json"
    if not log_path.exists():
        continue
    with open(log_path) as f:
        log = json.load(f)
    all_actions.extend(log["preprocessed_logs"])

all_actions = np.array(all_actions)
dim_names = ['X (前后)', 'Y (左右)', 'Z (上下)', 'Roll', 'Pitch', 'Yaw (偏航)']
used_dims = [0, 1, 2, 5]  # 训练用的 4 维

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()
for i, dim_idx in enumerate(used_dims):
    data = all_actions[:, dim_idx]
    axes[i].hist(data, bins=30, color='steelblue', alpha=0.7, edgecolor='white')
    axes[i].axvline(np.percentile(data, 1), color='red', linestyle='--', label='1st/99th')
    axes[i].axvline(np.percentile(data, 99), color='red', linestyle='--')
    axes[i].set_title(dim_names[dim_idx])
    axes[i].legend(fontsize=9)

fig.suptitle('AirSim 采集数据 - 动作分布 (训练用 4 维)', fontsize=14)
plt.tight_layout()
plt.show()

print(f"\n采集数据统计:")
print(f"  总帧数: {len(all_actions)}")
for i, dim_idx in enumerate(used_dims):
    data = all_actions[:, dim_idx]
    print(f"  {dim_names[dim_idx]:>12s}: mean={np.mean(data):+.4f}  std={np.std(data):.4f}  "
          f"1%={np.percentile(data, 1):+.4f}  99%={np.percentile(data, 99):+.4f}")
````
