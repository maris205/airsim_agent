## 8.2.1 RSSM：循环状态空间模型

RSSM 是 DreamerV3 世界模型的核心组件。它巧妙地融合了两种互补的建模方式：

### 两个"半脑"的融合

| 组件 | 类比 | 作用 | 来源 |
|------|------|------|------|
| 确定性状态 $h_t$ | 记事本 | 精准记录历史关键事件 | RNN/GRU |
| 随机性状态 $s_t$ | 骰子 | 表示对未来的不确定性 | 高斯分布采样 |

为什么需要两种状态？

- **只有确定性状态**（纯RNN）：无法表达"我不确定前方是墙还是门"这种不确定性
- **只有随机性状态**（纯SSM）：没有记忆，无法利用"3秒前我看到了一面墙"这种历史信息
- **两者结合**（RSSM）：既有记忆，又能表达不确定性

### RSSM 的工作流程

```
输入: 隐特征 z_t + 上一步动作 a_{t-1}
  ↓
GRU 更新确定性状态: h_t = GRU(h_{t-1}, [s_{t-1}, a_{t-1}])
  ↓
先验预测: p(s_t | h_t)          ← 仅凭记忆猜测
后验修正: q(s_t | h_t, z_t)     ← 结合实际观测修正
  ↓
完整状态: [h_t, s_t] → 用于预测奖励、图像等
```

![RSSM 直觉图](figures/rssm_intuition.png)

*图 8-4：RSSM 将"确定性记忆"与"随机性不确定性"融合，比纯 RNN 或纯概率模型更适合真实物理环境*

## 8.2.2 DreamerV3 四大组件

DreamerV3 由四个模块组成，形成一个完整的"学习-想象-决策"闭环：

### 1. 视觉编码器 (Encoder)
- 输入：原始图像 $o_t$（如 64×64 RGB）
- 输出：隐特征 $z_t$（如 32维向量）
- 作用：将高维图像压缩至低维隐空间

### 2. 世界模型 (World Model / RSSM)
- 输入：隐特征 $z_t$ + 动作 $a_t$
- 输出：预测的下一状态 $s_{t+1}$、预测奖励 $r_t$
- 作用：在隐空间内模拟环境动态

### 3. 策略网络 (Actor / Policy Network)
- 输入：当前状态 $[h_t, s_t]$
- 输出：最优动作 $a_t$
- 作用：基于世界模型的"想象"选择最优行动

### 4. 价值网络 (Critic / Value Network)
- 输入：当前状态 $[h_t, s_t]$
- 输出：价值估计 $V(s_t)$
- 作用：评估当前状态的长期价值，指导策略优化

### 训练闭环

```
真实环境交互 → 收集数据 (o, a, r, o') → 训练世界模型
                                              ↓
                                        在想象中展开轨迹
                                              ↓
                                    训练 Actor + Critic
                                              ↓
                                    Actor 输出动作 → 回到真实环境
```

## 8.2.3 DreamerV3 的关键创新

DreamerV3 相比前代（DreamerV1/V2）有几个重要改进：

### 1. Symlog 变换

奖励值的范围可能差异巨大（悬停任务奖励 ~1，碰撞惩罚 -100）。Symlog 变换将大数值压缩、小数值保留，让网络更容易学习：

$$\text{symlog}(x) = \text{sign}(x) \cdot \ln(|x| + 1)$$

### 2. 离散化回归

传统方法直接预测连续值（如奖励=3.7），DreamerV3 将其转化为分类问题：将值域离散化为多个区间，预测落在每个区间的概率。这让训练更稳定。

### 3. 无需任务特定调参

DreamerV3 的一大卖点是"一套超参数通吃所有任务"——从 Atari 游戏到机器人控制，不需要针对每个任务调整学习率、损失权重等超参数。这对教学和实际应用都非常友好。

## 8.2.4 动手实现：简化版 RSSM

下面我们用 PyTorch 实现一个简化版的 RSSM，观察确定性状态和随机性状态的行为。

````python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt

class SimpleRSSM(nn.Module):
    """简化版 RSSM：GRU(确定性) + 高斯采样(随机性)"""

    def __init__(self, obs_dim=32, action_dim=4, h_dim=64, s_dim=32):
        super().__init__()
        self.h_dim = h_dim
        self.s_dim = s_dim
        self.gru = nn.GRUCell(s_dim + action_dim, h_dim)
        # 先验网络：仅凭记忆预测
        self.prior_net = nn.Sequential(
            nn.Linear(h_dim, 64), nn.ReLU(),
            nn.Linear(64, s_dim * 2),  # 输出 mean + logstd
        )
        # 后验网络：结合观测修正
        self.posterior_net = nn.Sequential(
            nn.Linear(h_dim + obs_dim, 64), nn.ReLU(),
            nn.Linear(64, s_dim * 2),
        )

    def step(self, h, s, action, obs):
        # 1. GRU 更新确定性状态
        x = torch.cat([s, action], dim=-1)
        h_new = self.gru(x, h)

        # 2. 先验预测（仅凭记忆）
        prior_params = self.prior_net(h_new)
        prior_mean, prior_logstd = prior_params.chunk(2, dim=-1)
        prior_std = torch.exp(prior_logstd.clamp(-5, 2))

        # 3. 后验修正（结合观测）
        post_params = self.posterior_net(torch.cat([h_new, obs], dim=-1))
        post_mean, post_logstd = post_params.chunk(2, dim=-1)
        post_std = torch.exp(post_logstd.clamp(-5, 2))

        # 4. 从后验分布采样
        s_new = post_mean + post_std * torch.randn_like(post_std)

        return h_new, s_new, {
            "prior_mean": prior_mean, "prior_std": prior_std,
            "post_mean": post_mean, "post_std": post_std,
        }

rssm = SimpleRSSM()
print(f"RSSM 参数量: {sum(p.numel() for p in rssm.parameters()):,}")
print(f"确定性状态 h 维度: {rssm.h_dim}")
print(f"随机性状态 s 维度: {rssm.s_dim}")
````

````python
# 观察 RSSM 的行为：先验 vs 后验
rssm.eval()
with torch.no_grad():
    h = torch.zeros(1, 64)
    s = torch.zeros(1, 32)

    prior_stds = []
    post_stds = []

    for t in range(30):
        action = torch.randn(1, 4) * 0.3
        obs = torch.randn(1, 32)  # 模拟观测
        h, s, info = rssm.step(h, s, action, obs)
        prior_stds.append(info["prior_std"].mean().item())
        post_stds.append(info["post_std"].mean().item())

plt.figure(figsize=(8, 3))
plt.plot(prior_stds, 'r-o', markersize=3, label='先验不确定性（仅凭记忆）')
plt.plot(post_stds, 'b-o', markersize=3, label='后验不确定性（结合观测）')
plt.xlabel('时间步')
plt.ylabel('平均标准差')
plt.title('RSSM 先验 vs 后验不确定性')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print("先验（红色）通常比后验（蓝色）不确定性更高——因为观测提供了额外信息")
````

## 8.2.5 DreamerV3 与 AirSim 的数据链路

在后续实验中，DreamerV3 与 AirSim 的交互形成一个闭环：

```
┌─────────────────────────────────────────────────────┐
│                    训练闭环                          │
│                                                     │
│  AirSim ──(o,a,r,o')──→ 经验回放池                  │
│    ↑                         ↓                      │
│    │                    训练世界模型                  │
│    │                         ↓                      │
│    │                    想象中展开轨迹                │
│    │                         ↓                      │
│    │                    训练 Actor-Critic            │
│    │                         ↓                      │
│    └────── Actor输出动作 ─────┘                      │
│                                                     │
└─────────────────────────────────────────────────────┘
```

### 数据格式

每一步交互产生一个元组 `(observation, action, reward, next_observation, done)`：

| 字段 | 来源 | 维度 |
|------|------|------|
| observation | AirSim 相机 + IMU | 图像 64×64×3 + 状态 6维 |
| action | 策略网络输出 | 4维 (pitch, roll, yaw_rate, throttle) |
| reward | 奖励函数计算 | 标量 |
| done | 碰撞/超时检测 | 布尔值 |

![DreamerV3 训练闭环](figures/dreamerv3_training_loop.png)

*图 8-6：DreamerV3 训练闭环——真实交互提供数据，世界模型提供想象空间，Actor-Critic 在想象中优化*

## 8.2.6 小结

本节的核心要点：

1. **RSSM** 融合了 RNN 的记忆能力和概率模型的不确定性表达
2. DreamerV3 由**编码器、世界模型、策略网络、价值网络**四部分组成
3. 关键创新：symlog 变换、离散化回归、无需任务特定调参
4. 先验分布表示"仅凭记忆的猜测"，后验分布表示"结合观测的修正"
5. 与 AirSim 的数据链路形成"交互→训练→想象→决策"的闭环
